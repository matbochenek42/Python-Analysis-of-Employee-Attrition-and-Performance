# DATA CLEANING

### Duplicates removal (not needed though)

```python
df = df.drop_duplicates()
```

### Index set up

```python
df.set_index("EmployeeNumber", inplace=True)
```

### Using function to change the datatype to boolean


```python
def boolean(row, column_name):
    if row[column_name] == "No":
        return False
    else:
        return True

df["Attrition"] = df.apply(lambda row: boolean(row, "Attrition"), axis=1)
df["OverTime"]  = df.apply(lambda row: boolean(row, "OverTime"), axis=1)

```

### Values replacement within row

```python
df.BusinessTravel.unique()
```

```python
df["BusinessTravel"] = df["BusinessTravel"].str.replace("_", " ", regex=False)
```

### Dropping unnecessary column (same values in each rows)

```python
df.drop(labels = ["EmployeeCount", "Over18", "StandardHours"], axis=1, inplace=True) 
```
