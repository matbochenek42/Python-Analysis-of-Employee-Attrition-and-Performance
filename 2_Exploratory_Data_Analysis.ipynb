## INTRODUCTION

The goal of this file is to quickly inspect the data through visualization and understand the context before performing more complex analyses. It dives into correlations, gender and age distributions, employees’ education fields, and more.

```python
# CHANGE USER_PATH

from pathlib import Path


USER_PATH = Path(r"C:\Users\Mateusz\Desktop\ZBIORCZE\TECH\ANALIZA DANYCH\Python\Projekty MB") 


df_path = USER_PATH / "Python Analysis of Employee Attrition and Performance" / "dataframe.csv"
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv(df_path)
```

### Correlation between columns

```python
correlation = [
    'Age',
    'DistanceFromHome',
    'MonthlyIncome',
    'JobLevel',
    'TotalWorkingYears',
    'YearsAtCompany',
    'YearsInCurrentRole',
    'YearsSinceLastPromotion',
    'YearsWithCurrManager',
    'NumCompaniesWorked',
    'TrainingTimesLastYear',
    'EnvironmentSatisfaction',
    'JobSatisfaction',
    'WorkLifeBalance'
]
```

```python
df[correlation].corr()

sns.heatmap(df[correlation].corr(), annot = True)

plt.rcParams['figure.figsize'] = (20,20)

plt.show()
```

### Gender Distribution & Employees’ Marital Status

```python
gender = df["Gender"].value_counts()
status = df["MaritalStatus"].value_counts()


male_status = df[["Gender", "MaritalStatus"]].copy()
female_status = df[["Gender", "MaritalStatus"]].copy()

male_status = male_status[male_status["Gender"] == "Male"]
female_status = female_status[female_status["Gender"] == "Female"]

male_status = male_status.value_counts()
female_status = female_status.value_counts()

```

```python
sns.set_theme(style="ticks")

plt.figure(figsize=(12, 10))

plt.subplot(2, 2, 1)
plt.pie(gender, labels=["Male", "Female"], autopct='%1.0f%%', startangle=90)
plt.title("Gender Distribution of Employees")

plt.subplot(2, 2, 2)
plt.pie(status, labels=["Married", "Single", "Divorced"], autopct='%1.0f%%', startangle=90)
plt.title("Employees’ Marital Status")

plt.subplot(2, 2, 3)
plt.pie(male_status, labels=["Married", "Single", "Divorced"], autopct='%1.0f%%', startangle=90)
plt.title("Male Employees’ Marital Status")

plt.subplot(2, 2, 4)
plt.pie(female_status, labels=["Married", "Single", "Divorced"], autopct='%1.0f%%', startangle=90)
plt.title("Female Employees’ Marital Status")

#plt.tight_layout()
plt.suptitle("Gender Distribution & Employees’ Marital Status")
plt.show()
```

### Age Distribution

```python
plt.figure(figsize=(6, 6))

sns.set_theme(style="ticks")

df["Age"].plot(kind="hist", bins=20, edgecolor="black")

plt.title("Age Distribution of Employees")
plt.xlabel("Employee Age")
plt.ylabel("Number of Employees")

plt.show()
```

### Most Popular Education Fields, Job Roles and Department Among Employees

```python
education = df["EducationField"].value_counts().to_frame()

plt.figure(figsize=(6, 6))

sns.set_theme(style="ticks")

sns.barplot(data=education, x="count", y="EducationField", hue="count", palette="dark:b_r", legend=False)

sns.despine()
plt.title("Most Popular Fields of Education")
plt.xlabel("Number of Employees")
plt.ylabel("")
plt.show()
```

```python
job_roles = df["JobRole"].value_counts().to_frame()

plt.figure(figsize=(6, 6))

sns.set_theme(style="ticks")

sns.barplot(data=job_roles, x="count", y="JobRole", hue="count", palette="dark:b_r", legend=False)

sns.despine()
plt.title("Most Popular Job Roles")
plt.xlabel("Number of Employees")
plt.ylabel("")
plt.show()
```

```python
department = df["Department"].value_counts().to_frame()

plt.figure(figsize=(4, 4))

sns.set_theme(style="ticks")

sns.barplot(data=department, x="count", y="Department", hue="count", palette="dark:b_r", legend=False)

sns.despine()
plt.title("Most Popular Department")
plt.xlabel("Number of Employees")
plt.ylabel("")
plt.show()
```
